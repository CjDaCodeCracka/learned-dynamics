```julia
using Pkg
using Plots
include("methods.jl")
include("functions.jl")

data = euler(exp_decay,1,0,5,0.01)
#x(t) over time
plot1 = plot(data.times, 
    data.states, 
    xlabel = "Time", 
    ylabel = "x(t)", 
    label = "x(t)(Euler's Method)",
    title = "Exp Decay with Euler's Method")
#dx/dt over time
plot2 = plot(data.times, 
    data.derivatives, 
    xlabel = "Time", 
    ylabel = "dx/dt", 
    label = "dx/dt",
    title = "Rate of Change")
#dx/dt with respectt to x
plot3 = plot(data.states, 
    data.derivatives, 
    xlabel = "x(t)", 
    ylabel = "dx/dt", 
    label = "dx/dt with respect to x",
    title = "Slope Field")
exact = exp.(-2 .* data.times) #actual e^-2t over time
plot!(plot1, data.times, exact,
    label="Exact solution",
    linestyle=:dash)
plot4 = plot(data.times,
        abs.(exact-data.states),
        xlabel = "Time",
        ylabel = "exact vs euler",
        label = "error",
        title = "Error")
# display(plot1)
# display(plot2)
# display(plot3)
# display(plot4)
```
